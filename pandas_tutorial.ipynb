## Pickling
All pandas objects are equipped with to_pickle methods which use python's cPickle module to save data structires to disk using the pickel format.

1. What is Pickle?
Pickle is a Python format used to save Python objects into a file and load them back later.

For example, suppose you have a DataFrame with 1,000 rows. Instead of reading and processing the original CSV file every time, you can save the DataFrame as a Pickle file and load it whenever you need it.

```python
df_excel.to_pickle('Excel_Sample_pickle')  #to_pickle to save the dataframe as a pickle file
```

```python
df=pd.read_pickle('Excel_Sample_pickle')  #read_pickle to read the pickle file
```
