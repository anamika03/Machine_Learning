```python
data1 = ('index,a,b,c\n'
         '4,apple,bat,5.7\n'
         '8,orange,cow,10\n'
         '12,grape,dog,15\n') 
```

```python
pd.read_csv(StringIO(data1), index_col=0)  #index_col=0 to set the first column as index
```

```python
data2 = ('a,b,c\n'
         '1,apple,bat,\n'
         '2,orange,cow,\n'
         '3,grape,dog,\n')   
```

```python
pd.read_csv(StringIO(data2), index_col=False)  #index_col=False to not set any column as index
```

```python
## Qouting and Escape Characters. Very useful in NLP.

data3 = 'a,b\n"hello, \\"Bob\\", nice to see you",5\n'
```

```python
pd.read_csv(StringIO(data3),escapechar='\\')  #escapechar to escape the special characters in the data
```
