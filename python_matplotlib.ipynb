## Matplotlib 
Matplotlib is a plotting library for the python programming languages and ita numerical mathematics extension NumPy. It provides an object-oriented API for embedding plots into applications using general-purpose GUI toolkits like Tkinter, wxPython, QT or GKT+.

Some of the major Pros are:
- Generally easy to get started for simple plots.
- Support for custom labels and texts.
- Great control of every elements in a figure.
- High-quality output in many formats.
- Very customizable in general.

```python
import matplotlib.pyplot as plt

%matplotlib inline
```

```python
import numpy as np
```

```python
#Sample example

x = np.arange(0, 10)
y = np.arange(11, 21)
```

```python
a = np.arange(40, 50)
b = np.arange(50, 60)
```

```python
## Plotting using matplotlib
## plt scatter 

plt.scatter(x, y, c='g')
plt.xlabel("X axis")
plt.ylabel("Y axis")
plt.title("Scatter Graph in 2D")
plt.savefig("Scatter_Graph.png")
```

```python
## plt plot
y = x * x
plt.plot(x, y, 'ro--')
plt.xlabel("X axis")
plt.ylabel("Y axis")
plt.title("Plot Graph in 2D")
plt.savefig("Plot_Graph.png")
```

```python
## creating subplots

plt.subplot(2, 2, 1)   #(nrow, ncol, pos)
plt.plot(x,y,'r--')
plt.subplot(2, 2, 2)
plt.plot(x,y,'bo--')
plt.subplot(2, 2, 3)
plt.plot(a,b,'g*--')
plt.subplot(2, 2, 4)
plt.plot(a,b,'y-')
```

```python
x = np.arange(1, 11)
y = 2 * x + 7
plt.plot(x, y, 'ro--')
plt.xlabel("X axis")
plt.ylabel("Y axis")
plt.title("MatlibPlot Demo")
plt.show()
```

```python
# compute the x and y cordinates for points on a sine curve
x = np.arange(0, 4 * np.pi, 0.1)
y_sin = np.sin(x)

plt.subplot(2,1,1)
plt.title("Sine wave form")
plt.plot(x,y_sin, 'g--')


y_cos = np.cos(x)
plt.subplot(2,1,2)
plt.title("Cosine wave form")
plt.plot(x,y_cos, 'r--')

plt.show()
```

```python
## Bar Plot

x = [2, 4, 6]
y = [11, 16, 9]

x2 = [3, 8, 11]
y2 = [6, 15, 7]

plt.bar(x,y, color='r')
plt.bar(x2,y2, color='g')
plt.title("Bar Graph")
plt.xlabel("X axis")
plt.ylabel("Y axis")

plt.show()
```

## Histograms
A Histogram is a graph used to understand the distribution of numerical data.
It shows how frequently values occur within different ranges.
- A histogram divides numerical values into groups called bins. 
- If you don't provide the bins count in plt.hist(), Matplotlib automatically uses 10 bins by default.
- X-axis: Data ranges (bins).
- Y-axis: Frequency (number of observations).
- Bars: Show how many values fall within each range.


```python
marks = np.array([35, 40, 45, 48, 50, 52, 55, 56, 58, 60,
         62, 65, 67, 70, 72, 75, 80, 85, 90, 95])
plt.hist(marks, bins=10, edgecolor='black')

plt.xlabel("Marks")
plt.ylabel("Frequency")
plt.title("Distribution of Student Marks")

plt.show()
```

## Box Plot
A Box Plot (also called Box-and-Whisker Plot) is a statistical graph used to understand the distribution of numerical data. It helps us identify the median, spread of data, quartiles and outliers.

- Minimum	:                Lowest non-outlier value
- Q1 (First Quartile)	:    25% of data lies below this value
- Median (Q2):            Middle value; 50% of data lies below it
- Q3 (Third Quartile)	:    75% of data lies below this value
- Maximum	         :       Highest non-outlier value

```python
data = [np.random.normal(0, std, 100) for std in range(1, 4)]

# rectangular box plot
plt.boxplot(data, vert=True, patch_artist=True)
```

# Pie Chart

```python
labels = 'Python', 'C++', 'Ruby', 'Java'
sizes = [245, 130, 214, 210]
colours = ['gold', 'green', 'yellowgreen', 'orange']
explode = (0.1,0,0,0)   #explode 1st slide

#plot
plt.pie(sizes, explode=explode, labels=labels, colors=colours, autopct='%1.1f%%', shadow=True)

plt.axis('equal')
plt.savefig("pie_chart")
plt.show()
```
